```python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer

model_id = "Qwen/Qwen2.5-Coder-7B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(
    model_id, torch_dtype=torch.float16, device_map="auto"
)

# Crucial step: provide the raw data schema so the model doesn't hallucinate column names
messages = [
    {
        "role": "system",
        "content": "You are an expert PowerBI developer. Provide optimized DAX measures or Power Query M code based on the data schema provided."
    },
    {
        "role": "user",
        "content": """
        Raw Data Schema: Table name is 'SalesData'. Columns: Date (DateTime), Revenue (Numeric), Region (Text), Cost (Numeric).

        Task: Write a DAX measure to calculate a 3-month rolling average of net profit (Revenue - Cost).
        """
    }
]

inputs = tokenizer.apply_chat_template(messages, tokenize=True, add_generation_prompt=True, return_tensors="pt").to("cuda")
outputs = model.generate(inputs, max_new_tokens=400, temperature=0.1) # Low temperature ensures syntactically strict DAX

print(tokenizer.decode(outputs[0][inputs.shape[-1]:], skip_special_tokens=True))
```
